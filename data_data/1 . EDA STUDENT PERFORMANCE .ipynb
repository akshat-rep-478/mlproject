#### BIVARIATE ANALYSIS (  Is lunch type intake has any impact on student's performance ? )

```python
# 1. Create a single axis canvas
f, ax = plt.subplots(1, 1, figsize=(20, 8))

# 2. Plot the data (Use ax instead of ax[0], and x='col' instead of df['col'])
sns.countplot(x='parental_level_of_education', data=df, palette='bright', hue='lunch', saturation=0.95, ax=ax)

# 3. Add a clear title
ax.set_title('Parental Education vs Lunch Type', color='black', size=25, weight='bold')

# 4. Add the labels to the bars
for container in ax.containers:
    ax.bar_label(container, color='black', size=20, padding=3)

plt.show()

# 2. Generate and print the statistical summary
lunch_stats = df.groupby('lunch')[['math_score', 'reading_score', 'writing_score', 'average']].mean().round(2)
print("Summary Statistics by Lunch Type:")
print(lunch_stats)

# 3. Set up the visualization canvas and style
plt.figure(figsize=(14, 6))
sns.set_context("talk")

# 4. Melt the DataFrame
# This reshapes the data so that 'subject' becomes a single column, 
# making it easy to plot all four metrics side-by-side using Seaborn's 'hue' argument.
df_melted = pd.melt(df, id_vars=['lunch'], 
                    value_vars=['math_score', 'reading_score', 'writing_score', 'average'], 
                    var_name='subject', value_name='score')

# 5. Plot the grouped bar chart
ax = sns.barplot(x='subject', y='score', hue='lunch', data=df_melted, palette='Set2')

# 6. Add exact values to the top of each bar
for container in ax.containers:
    ax.bar_label(container, fmt='%.1f', padding=3, size=12)

# 7. Format labels, title, and legend
plt.title('Student Performance vs Lunch Uptake', weight='bold')
plt.ylabel('Average Score')
plt.xlabel('Subject')
plt.legend(title='Lunch Type', bbox_to_anchor=(1.05, 1), loc='upper left')

# 8. Adjust layout and display
plt.tight_layout()
plt.show()
```

#### Insights 
- Students who get Standard Lunch tend to perform better than students who got free/reduced lunch
